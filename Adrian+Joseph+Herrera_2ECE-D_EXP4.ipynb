### Overview & Objectives
This notebook demonstrates data wrangling and visualization techniques using `pandas`, `matplotlib`, and `seaborn` on an ECE Board Exam dataset.

**Key Learning Goals:**
1. Multi-condition filtering using Pandas boolean indexing.
2. Creating tailored DataFrames without altering the source dataset.
3. Aggregating numerical values across categorical groupings using `.groupby()`.
4. Generating readable multi-panel visualizations with proper labeling and consistent scales.


## Step 1: Initial Setup and Dataset Loading

We import the required libraries, load `board2.xlsx` into `df_raw`, and compute the `Average` column across all four exam subjects (`Math`, `Electronics`, `GEAS`, `Communication`).


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")

# Load raw dataset
df_raw = pd.read_excel('board2.xlsx')

# Compute the missing 'Average' column across the 4 subject areas
df_raw['Average'] = df_raw[['Math', 'Electronics', 'GEAS', 'Communication']].mean(axis=1)

# Display first few rows to verify structure
df_raw.head()
```

---
## Problem A: Visayas Communication DataFrame

### Task Objective:
Filter for students whose `Hometown` is **Visayas** AND whose `Track` is **Communication**.

### Implementation Details:
1. **Explicit Multi-Condition Filtering:** Applied using `(df_raw['Hometown'] == 'Visayas') & (df_raw['Track'] == 'Communication')`.
2. **Column Selection:** Retaining only `['Name', 'Gender', 'Math', 'Electronics', 'Average']` in exact order.
3. **Display:** Outputting the DataFrame and total row count.


```python
# Apply explicit boolean conditions
vis_comm_mask = (df_raw['Hometown'] == 'Visayas') & (df_raw['Track'] == 'Communication')
VisComm = df_raw.loc[vis_comm_mask, ['Name', 'Gender', 'Math', 'Electronics', 'Average']].copy()

# Display results
print("--- Visayas Communication DataFrame ---")
display(VisComm)

print(f"\nTotal rows in VisComm: {len(VisComm)}")
```

---
## Problem B: Visayas Female DataFrame

### Task Objective:
1. Filter students whose `Hometown` is **Visayas** AND `Gender` is **Female**.
2. Retain columns: `Name`, `Track`, `GEAS`, `Electronics`, `Average`.
3. Display `VisFemale`.
4. Filter and display rows from `VisFemale` where `Average >= 60` without modifying `VisFemale`.

```python
# Create VisFemale DataFrame
vis_female_mask = (df_raw['Hometown'] == 'Visayas') & (df_raw['Gender'] == 'Female')
VisFemale = df_raw.loc[vis_female_mask, ['Name', 'Track', 'GEAS', 'Electronics', 'Average']].copy()

# 1. Display full VisFemale DataFrame
print("--- Full VisFemale DataFrame ---")
display(VisFemale)

# 2. Display sub-filtered view (Average >= 60) without overwriting VisFemale
print("\n--- VisFemale Students with Average >= 60 ---")
display(VisFemale[VisFemale['Average'] >= 60])
```

---
## Problem C: Category-Average Visualization & Analysis

### Task Objective:
Calculate category means for `Track`, `Gender`, and `Hometown`, display summary tables, and generate a 3-panel bar chart comparing mean exam averages.


```python
# Compute category means using groupby
mean_track = df_raw.groupby('Track')['Average'].mean().reset_index()
mean_gender = df_raw.groupby('Gender')['Average'].mean().reset_index()
mean_hometown = df_raw.groupby('Hometown')['Average'].mean().reset_index()

# Display summary tables
print("=== Mean Average by Track ===")
display(mean_track)

print("\n=== Mean Average by Gender ===")
display(mean_gender)

print("\n=== Mean Average by Hometown ===")
display(mean_hometown)
```

```python
# Create 3-panel figure with a shared y-axis scale
fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharey=True)

# Plot 1: Track
sns.barplot(data=mean_track, x='Track', y='Average', ax=axes[0], palette='Blues_d')
axes[0].set_title('Mean Average by Track', fontweight='bold', fontsize=12)
axes[0].set_ylabel('Mean Board Exam Average', fontsize=10)
for p in axes[0].patches:
    axes[0].annotate(f"{p.get_height():.2f}", (p.get_x() + p.get_width() / 2., p.get_height()), 
                     ha='center', va='center', xytext=(0, 5), textcoords='offset points')

# Plot 2: Gender
sns.barplot(data=mean_gender, x='Gender', y='Average', ax=axes[1], palette='Purples_d')
axes[1].set_title('Mean Average by Gender', fontweight='bold', fontsize=12)
axes[1].set_ylabel('')
for p in axes[1].patches:
    axes[1].annotate(f"{p.get_height():.2f}", (p.get_x() + p.get_width() / 2., p.get_height()), 
                     ha='center', va='center', xytext=(0, 5), textcoords='offset points')

# Plot 3: Hometown
sns.barplot(data=mean_hometown, x='Hometown', y='Average', ax=axes[2], palette='Greens_d')
axes[2].set_title('Mean Average by Hometown', fontweight='bold', fontsize=12)
axes[2].set_ylabel('')
for p in axes[2].patches:
    axes[2].annotate(f"{p.get_height():.2f}", (p.get_x() + p.get_width() / 2., p.get_height()), 
                     ha='center', va='center', xytext=(0, 5), textcoords='offset points')

plt.suptitle('Comparison of Board Exam Performance Across Categorical Groups', fontsize=14, fontweight='bold', y=1.03)
plt.tight_layout()
plt.show()
```

### Data Interpretation Statements

1. **Track:** Based on the observed dataset, students in the **Communication** track achieved the highest sample mean board-exam score in `Average` (67.98).
2. **Gender:** Among the recorded student entries, **Male** students achieved the highest sample mean board-exam score in `Average` (67.18).
3. **Hometown:** Students originating from **Luzon** recorded the highest sample mean board-exam score in `Average` (68.08).

> **Methodological Note:** These observations describe the sample group only and do not establish causal relationships between student demographics/track choices and board exam scores.
